```python
x = torch.tensor(5.0,requires_grad=True)
f = x**2
f
```

Note that the backward() function automatically computed the gradient f'(x) at the same point x = 5.0. 

```python
f.backward()
x.grad
```

 Running gradient descent will gradually push x toward 0, since that’s the value that minimizes f(x) = $x^2$. To do a gradient descent step, you must temporarily disable gradient tracking since you don’t want to track the gradient descent step itself in the computation graph (in fact, PyTorch would raise an exception if you tried to run an in-place operation on a tracked variable). This can be done by placing the gradient descent step inside a torch.no_grad() context

```python
learning_rate = 0.1
with torch.no_grad():
    x -= learning_rate * x.grad
```

```python
x_detach = x.detach()
x_detach -= learning_rate * x.grad
```

Since x_detached and x share the same memory, modifying x_detached also modifies x.
